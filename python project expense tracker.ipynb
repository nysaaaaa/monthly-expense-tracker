```python
expenses = []

user_name = ""
selected_month = ""
monthly_budget = 0.0




def show_title():

    print()
    print("=" * 65)
    print("                 MONTHLY EXPENSE TRACKER")
    print("=" * 65)
    print()




def welcome_message():

    print("Welcome to the Monthly Expense Tracker!")
    print("This program helps you manage your monthly spending.")
    print()



def get_user_details():

    global user_name
    global selected_month

    print("-" * 65)
    print("                    USER DETAILS")
    print("-" * 65)

    user_name = input("Enter your name: ")

    selected_month = input(
        "Enter the month you are tracking: "
    )

    print()
    print("Hello,", user_name + "!")
    print("You are tracking expenses for:", selected_month)
    print()



def set_budget():

    global monthly_budget

    print("-" * 65)
    print("                    SET MONTHLY BUDGET")
    print("-" * 65)

    while True:

        try:

            amount = input("Enter your monthly budget: $")

            amount = amount.replace("$", "")

            new_budget = float(amount)

            if new_budget < 0:

                print("Budget cannot be negative.")
                print("Please try again.")

            else:

                monthly_budget = new_budget

                print()
                print("Monthly budget set successfully!")
                print("Budget: $", monthly_budget)

                break

        except ValueError:

            print("Invalid amount.")
            print("Please enter a number.")




def show_menu():

    print()
    print("=" * 65)
    print("                         MAIN MENU")
    print("=" * 65)

    print("1.  Add Expense")
    print("2.  View All Expenses")
    print("3.  View Total Expenses")
    print("4.  View Budget Status")
    print("5.  View Highest Expense")
    print("6.  View Lowest Expense")
    print("7.  View Expenses by Category")
    print("8.  Search Expense by Date")
    print("9.  Edit Expense")
    print("10. Delete Expense")
    print("11. Set New Budget")
    print("12. View Monthly Summary")
    print("13. View User Details")
    print("14. Exit")

    print("=" * 65)



def get_amount():

    while True:

        try:

            amount = input("Enter amount: $")

            amount = amount.replace("$", "")

            amount = float(amount)

            if amount < 0:

                print("Amount cannot be negative.")

            else:

                return amount

        except ValueError:

            print("Please enter a valid number.")



def add_one_expense():

    print()
    print("-" * 65)
    print("                      ADD EXPENSE")
    print("-" * 65)

    date = input("Enter date (DD-MM-YYYY): ")

    category = input("Enter category: ")

    description = input("Enter description: ")

    amount = get_amount()

    expense = {
        "date": date,
        "category": category,
        "description": description,
        "amount": amount
    }

    expenses.append(expense)

    print()
    print("Expense added successfully!")
    print()
    print("Date:", date)
    print("Category:", category)
    print("Description:", description)
    print("Amount: $", amount)

    check_budget()



def add_expenses():

    while True:

        add_one_expense()

        print()
        print("Do you want to add another expense?")
        print("1. Yes")
        print("2. No")

        choice = input("Enter your choice: ")

        if choice == "1":

            print()
            print("Adding another expense...")
            print()

        elif choice == "2":

            print()
            print("Returning to main menu.")
            break

        else:

            print()
            print("Invalid choice.")
            print("Returning to main menu.")
            break




def view_all_expenses():

    print()
    print("-" * 65)
    print("                    ALL EXPENSES")
    print("-" * 65)

    if len(expenses) == 0:

        print("No expenses have been added yet.")
        return

    for i in range(len(expenses)):

        print()
        print("Expense", i + 1)
        print("-" * 35)

        print("Date:", expenses[i]["date"])

        print("Category:", expenses[i]["category"])

        print("Description:", expenses[i]["description"])

        print("Amount: $", expenses[i]["amount"])

    print()
    print("-" * 65)
    print("Total number of expenses:", len(expenses))



def calculate_total():

    total = 0.0

    for expense in expenses:

        total = total + expense["amount"]

    return total




def view_total():

    print()
    print("-" * 65)
    print("                    TOTAL EXPENSES")
    print("-" * 65)

    if len(expenses) == 0:

        print("No expenses recorded.")
        return

    total = calculate_total()

    print("Month:", selected_month)
    print("Total expenses: $", total)

    if monthly_budget > 0:

        print("Monthly budget: $", monthly_budget)

        if total > monthly_budget:

            print()
            print("WARNING!")
            print("You have exceeded your monthly budget.")

        elif total == monthly_budget:

            print()
            print("You have reached your monthly budget.")

        else:

            remaining = monthly_budget - total

            print("Remaining budget: $", remaining)




def check_budget():

    if monthly_budget <= 0:

        return

    total = calculate_total()

    print()

    if total > monthly_budget:

        print("!" * 65)
        print("                 WARNING: BUDGET EXCEEDED")
        print("!" * 65)

        print("Your budget: $", monthly_budget)

        print("Your spending: $", total)

        print(
            "Amount over budget: $",
            total - monthly_budget
        )

        print("!" * 65)

    elif total == monthly_budget:

        print("You have reached your budget exactly.")

    else:

        remaining = monthly_budget - total

        print("Budget remaining: $", remaining)




def budget_status():

    print()
    print("-" * 65)
    print("                    BUDGET STATUS")
    print("-" * 65)

    if monthly_budget <= 0:

        print("No monthly budget has been set.")
        return

    total = calculate_total()

    remaining = monthly_budget - total

    print("Monthly budget: $", monthly_budget)

    print("Total spending: $", total)

    print()

    if remaining > 0:

        print("Status: Within Budget")

        print("Remaining: $", remaining)

    elif remaining == 0:

        print("Status: Budget Reached")

        print("You have used your entire budget.")

    else:

        print("Status: OVER BUDGET")

        print("Amount over budget: $", abs(remaining))




def highest_expense():

    print()
    print("-" * 65)
    print("                    HIGHEST EXPENSE")
    print("-" * 65)

    if len(expenses) == 0:

        print("No expenses recorded.")

        return

    highest = expenses[0]

    for expense in expenses:

        if expense["amount"] > highest["amount"]:

            highest = expense

    print("Date:", highest["date"])

    print("Category:", highest["category"])

    print("Description:", highest["description"])

    print("Amount: $", highest["amount"])




def lowest_expense():

    print()
    print("-" * 65)
    print("                    LOWEST EXPENSE")
    print("-" * 65)

    if len(expenses) == 0:

        print("No expenses recorded.")

        return

    lowest = expenses[0]

    for expense in expenses:

        if expense["amount"] < lowest["amount"]:

            lowest = expense

    print("Date:", lowest["date"])

    print("Category:", lowest["category"])

    print("Description:", lowest["description"])

    print("Amount: $", lowest["amount"])



def category_search():

    print()
    print("-" * 65)
    print("                EXPENSES BY CATEGORY")
    print("-" * 65)

    if len(expenses) == 0:

        print("No expenses recorded.")

        return

    category = input("Enter category: ")

    found = False

    category_total = 0.0

    for expense in expenses:

        if expense["category"].lower() == category.lower():

            found = True

            print()
            print("Date:", expense["date"])

            print("Category:", expense["category"])

            print("Description:", expense["description"])

            print("Amount: $", expense["amount"])

            category_total = (
                category_total
                + expense["amount"]
            )

    if found:

        print()
        print("Total spent on", category, ": $", category_total)

    else:

        print()
        print("No expenses found in this category.")



def date_search():

    print()
    print("-" * 65)
    print("                  SEARCH BY DATE")
    print("-" * 65)

    if len(expenses) == 0:

        print("No expenses recorded.")

        return

    search_date = input(
        "Enter date (DD-MM-YYYY): "
    )

    found = False

    for expense in expenses:

        if expense["date"] == search_date:

            found = True

            print()
            print("Category:", expense["category"])

            print("Description:", expense["description"])

            print("Amount: $", expense["amount"])

    if not found:

        print()
        print("No expenses found on this date.")




def edit_expense():

    print()
    print("-" * 65)
    print("                    EDIT EXPENSE")
    print("-" * 65)

    if len(expenses) == 0:

        print("No expenses available.")

        return

    view_all_expenses()

    while True:

        try:

            number = int(
                input(
                    "\nEnter expense number to edit: "
                )
            )

            if number >= 1 and number <= len(expenses):

                break

            else:

                print("Invalid expense number.")

        except ValueError:

            print("Please enter a number.")


    index = number - 1

    print()
    print("Enter new information.")

    new_date = input(
        "Enter new date (DD-MM-YYYY): "
    )

    new_category = input(
        "Enter new category: "
    )

    new_description = input(
        "Enter new description: "
    )

    new_amount = get_amount()

    expenses[index]["date"] = new_date

    expenses[index]["category"] = new_category

    expenses[index]["description"] = new_description

    expenses[index]["amount"] = new_amount

    print()
    print("Expense updated successfully!")

    check_budget()




def delete_expense():

    print()
    print("-" * 65)
    print("                   DELETE EXPENSE")
    print("-" * 65)

    if len(expenses) == 0:

        print("No expenses available.")

        return

    view_all_expenses()

    while True:

        try:

            number = int(
                input(
                    "\nEnter expense number to delete: "
                )
            )

            if number >= 1 and number <= len(expenses):

                break

            else:

                print("Invalid expense number.")

        except ValueError:

            print("Please enter a number.")


    index = number - 1

    print()
    print("Selected expense:")

    print(
        "Description:",
        expenses[index]["description"]
    )

    print(
        "Amount: $",
        expenses[index]["amount"]
    )

    confirm = input(
        "Delete this expense? (yes/no): "
    )

    if confirm.lower() == "yes":

        expenses.pop(index)

        print()
        print("Expense deleted successfully.")

    else:

        print()
        print("Expense was not deleted.")



def change_budget():

    print()
    print("-" * 65)
    print("                    CHANGE BUDGET")
    print("-" * 65)

    print("Current budget: $", monthly_budget)

    set_budget()

    check_budget()




def monthly_summary():

    print()
    print("=" * 65)
    print("                    MONTHLY SUMMARY")
    print("=" * 65)

    print("Name:", user_name)

    print("Month:", selected_month)

    print()

    if len(expenses) == 0:

        print("No expenses recorded.")

        return

    total = calculate_total()

    print("Number of expenses:", len(expenses))

    print("Total spent: $", total)

    print("Monthly budget: $", monthly_budget)

    if monthly_budget > 0:

        remaining = monthly_budget - total

        if remaining > 0:

            print("Remaining budget: $", remaining)

            print("Status: Within Budget")

        elif remaining == 0:

            print("Status: Budget Reached")

        else:

            print("Status: OVER BUDGET")

            print(
                "Amount over budget: $",
                abs(remaining)
            )

    highest = expenses[0]

    lowest = expenses[0]

    for expense in expenses:

        if expense["amount"] > highest["amount"]:

            highest = expense

        if expense["amount"] < lowest["amount"]:

            lowest = expense

    print()

    print(
        "Highest expense: $",
        highest["amount"]
    )

    print(
        "Highest expense description:",
        highest["description"]
    )

    print()

    print(
        "Lowest expense: $",
        lowest["amount"]
    )

    print(
        "Lowest expense description:",
        lowest["description"]
    )



def user_details():

    print()
    print("-" * 65)
    print("                    USER DETAILS")
    print("-" * 65)

    print("Name:", user_name)

    print("Month:", selected_month)

    print("Monthly budget: $", monthly_budget)

    print("Number of expenses:", len(expenses))




def process_menu(choice):

    if choice == "1":

        add_expenses()

    elif choice == "2":

        view_all_expenses()

    elif choice == "3":

        view_total()

    elif choice == "4":

        budget_status()

    elif choice == "5":

        highest_expense()

    elif choice == "6":

        lowest_expense()

    elif choice == "7":

        category_search()

    elif choice == "8":

        date_search()

    elif choice == "9":

        edit_expense()

    elif choice == "10":

        delete_expense()

    elif choice == "11":

        change_budget()

    elif choice == "12":

        monthly_summary()

    elif choice == "13":

        user_details()

    elif choice == "14":

        return False

    else:

        print()
        print("Invalid choice.")
        print("Please enter a number from 1 to 14.")

    return True




def main():

    show_title()

    welcome_message()

    get_user_details()

    print()
    print("-" * 65)
    print("                    BUDGET SETUP")
    print("-" * 65)

    set_budget()

    running = True

    while running:

        show_menu()

        choice = input(
            "Enter your choice (1-14): "
        )

        running = process_menu(choice)

    print()
    print("=" * 65)
    print("             THANK YOU FOR USING THE TRACKER")
    print("=" * 65)

    print("Name:", user_name)

    print("Month:", selected_month)

    print("Final spending: $", calculate_total())

    print("=" * 65)



main()
```
